### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
while True:
    try:
        sides = list(map(float, input("Input the side lengths of your triangle:").split()))
        if len(sides) != 3:
            print("A triangle has 3 sides only.")
        else:
            break
    except ValueError:
        print("Invalid user input. Please try again.")

longest_side = 0
circumference = 0
for x in sides:
    if x > longest_side:
        longest_side = x # find the longest side length
        circumference += x # compute the circumference


def area_of_triangle(sides): # function for computing the area
    half_perimeter = (sides[0] + sides[1] + sides[2]) / 2
    area = (half_perimeter * (half_perimeter - sides[0]) * (half_perimeter - sides[1]) * (half_perimeter - sides[2])) ** 0.5
    return area

if circumference <= longest_side * 2: # Short side A + Short side B smaller or equal Longest side => Can not form a triangle
    print("The value for the 3 sides length you provided cannot form a triangle.")
elif any(x < 0 for x in sides):
    print ("The length of a side cannot be a negative number")
else:
    print(f"The area of your triangle is {area_of_triangle(sides)}.")

```

### Problem 7
#### Write a function that returns True if a, b, c can be the side lengths of a triangle; otherwise return False.

```python
while True:
    try:
        sides = list(map(float, input("Input the side lengths of your triangle:").split()))
        if len(sides) != 3:
            print("A triangle has 3 sides only.")
        elif any(x < 0 for x in sides):
            print("The side length of a triangle cannot be a negative number.")
        else:
            break
    except ValueError:
        print("Invalid user input. Please try again.")
def ValidTriangleCheck(sides):
    longest_side = 0
    perimeter = 0
    for x in sides:
        if x > longest_side:
            longest_side = x
        perimeter += x
    if perimeter < longest_side * 2:
        return False
    else:
        return True
print(ValidTriangleCheck(sides))
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
while True:
    try:
        x = input("Choose a character in the alphabet")
        n = int(input("Choose your k value"))
        if not x.isalpha() or len(x) != 1 or n < 0 or n > 26:
           print("Invalid input.")
        else:
            break
    except ValueError:
        print("Invalid input.")

def CharacterOffset(x, n):
    if 97 <= ord(x) <= 122:
        shift_amount = n % 26
        new_character_lowercase = chr(ord(x) + shift_amount)
        return new_character_lowercase
    elif 65<= ord(x) <= 90:
        shift_amount = n % 26
        new_character_uppercase = chr(ord(x) + shift_amount)
        return new_character_uppercase
print(CharacterOffset(x, n))
    


```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
while True:
    try:
        x = input("Enter your letter:")
        if not x.isalpha() or len(x) != 1:
            raise ValueError
        else:
            break
    except ValueError:
        print("Invalid user input.")

def CharacterConverter(x):
    if 97 <= ord(x) <= 122:
        new_character_uppercase = chr(ord(x) - 32)
        return new_character_uppercase
    elif 65<= ord(x) <= 90:
        return "Already is an uppercase letter"
print(CharacterConverter(x))
```

### Problem 10
#### Write a function that, given a lowercase letter, returns a string containing that letter repeated 10 times.

```python
while True:
    try:
        x = input("Enter your letter:")
        if not x.isalpha() or len(x) != 1 or not x.islower():
            raise ValueError
        else:
            break
    except ValueError:
        print("Invalid user input.")

def RepeatCharacter(x):
    return x*10
print(RepeatCharacter(x))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
while True:
    try:
        n = int(input("Input the height of your triangle:"))
        break
    except ValueError:
        print("Invalid user input.")
for x in range(n): 
    print("*" * (x + 1)) #since range(n) starts from 0, we have to + 1 to x
```
